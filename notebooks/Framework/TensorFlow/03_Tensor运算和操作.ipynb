#### 3）张量比较
为了计算分类任务的准确率等指标，一般需要将预测结果和真实标签比较，统计比较结果中正确的数量来就是计算准确率。考虑 100 个样本的预测结果：

```python
out = tf.random.normal([100, 10])
out = tf.nn.softmax(out, axis = 1) # 输出转换为概率
pred = tf.argmax(out, axis = 1) # 选取预测值
pred
```

可以看到我们模拟的 100 个样本的预测值，我们与这 100 样本的真实值比较：

```python
y = tf.random.uniform([100], dtype = tf.int64, maxval = 10)
y
```

即可获得每个样本是否预测正确。通过 tf.equal(a, b)(或 tf.math.equal(a, b))函数可以比较这 2 个张量是否相等：

```python
out = tf.equal(pred, y)
out
```

tf.equal()函数返回布尔型的张量比较结果，只需要统计张量中 True 元素的个数，即可知道预测正确的个数。为了达到这个目的，我们先将布尔型转换为整形张量，再求和其中 1 的个数，可以得到比较结果中 True 元素的个数：

```python
out = tf.cast(out, dtype = tf.float32) # bool -> int
correct_sum = tf.reduce_sum(out) # True的个数
correct_mean = tf.reduce_mean(out) # True的概率
correct_sum, correct_mean
```

#### 4）Broadcasting机制
上面的所有实例中所用到的张量都是在维度数和形状相同情况下进行，那么，当两个张量维度数或者形状不一样时能不能进行运算呢?  
  
Broadcasting机制解除了只能维度数和形状相同的张量才能进行运算的限制，当两个数组进行算术运算时，TensorFlow的Broadcasting机制首先对维度较低的张量形状数组填充1，从后向前，逐元素比较两个数组的形状，当逐个比较的元素值（注意，这个元素值是指描述张量形状数组的值，不是张量的值）满足以下条件时，认为满足 Broadcasting 的条件：  
- （1）相等

- （2）其中一个张量形状数组元素值为1。

再举一些例子加深理解：

- A：(2d array): 5 x 4
- B：(1d array): 1
- Result：(2d array): 5 x 4
---
- A：(2d array): 5 x 4
- B：(1d array): 4
- Result：(2d array): 5 x 4
---
- A：(3d array): 15 x 3 x 5
- B：(3d array): 15 x 1 x 5
- Result：(3d array): 15 x 3 x 5
---
- A：(3d array): 15 x 3 x 5
- B：(2d array): 3 x 5
- Result：(3d array): 15 x 3 x 5
---
- A：(3d array): 15 x 3 x 5
- B：(2d array): 3 x 1
- Result：(3d array): 15 x 3 x 5

一些反例（不满足 Broadcasting 规则 ）：

- A (1d array): 3
- B (1d array): 4
---
- A (2d array): 8 x 4
- B (3d array): 8 x 4 x 3

```python
import numpy as np
a = tf.constant(1, dtype = tf.int64)
b = tf.constant(np.arange(12), shape=(3,4))
a + b
```

```python
a = tf.constant([1, 2, 3, 4], dtype = tf.int64)
b = tf.constant(np.arange(12), shape=(3,4))
a + b
```

```python
a = tf.constant([[1], [2], [3]], dtype = tf.int64)
b = tf.constant(np.arange(12), shape=(3,4))
a + b
```

```python
a = tf.constant(np.arange(8), shape=(2, 1, 4)) #(2,1，4)  -->(2,3，4)
b = tf.constant(np.arange(24), shape=(2, 3, 4))
a, b, a+b
```

#### 5）矩阵相乘
矩阵相乘前提条件:

左矩阵的列必须和右矩阵的行相同；即与M*K矩阵相乘的矩阵K*N，K必须相同，相乘以后的结果为M*N, 即 M*K 乘以 K*N = M*N。

比如：(3,2)@(2,4)->(3,4)



tensorflow中实现矩阵相乘的api有两个：

- a@b

- tf.matmul(a,b)

高维矩阵的矩阵乘法只会在矩阵的最后二个维度上进行，其他维度依然适用于广播机制，例如：

**A：2x2x2x3 B：3x2 两个矩阵如何实现矩阵相乘？**

- 首先B进行扩维，B：3x2--》1x1x3x2
- 然后B进行复制，保证跟A的前两个维度一致 B：1x1x3x2--》2x2x3x2
- 然后A和B在最后两个维度上进行矩阵相乘，前面的维度长度不变 2x2x2x3 @ 2x2x3x2 = 2x2x2x2



```python
import tensorflow as tf
a = tf.constant(np.arange(6), shape = (2, 3))
b = tf.constant(np.arange(6), shape = (3, 2))
a, b
```

```python
tf.matmul(a, b), a @ b
```

```python
a = tf.constant(np.arange(24), shape = (2, 2, 2, 3))
b = tf.constant(np.arange(6), shape = (3, 2))
a, b, a @ b
```

#### 2）部分排序
sort()方法和argsort()方法都是对给定Tensor的所有元素进行排序，在某些情况下如果我们只是要获取排序的前几个元素，这时候使用sort()或argsort()方法就有些浪费时间了，这时候可以使用top_k()方法。top_k()方法可以指定获取前k个元素。

**tf.math.top_k(a,k=1)**

- a:需要排序的tensor

- k：取前k个元素，默认k=1

**top_k()方法只能对最后一维度进行排序。**

```python
a, tf.math.top_k(a)
```

从上述输出可以看到，top_k()方法返回的是一个TopKV2类型对象，内部包含两部分数据：

- 第一部分是排序后的真实数据[5, 4]，可以通过TopKV2对象的values属性获取；

- 第二部分是排序后数据所在原Tensor中的索引[2, 5]，可以通过TopKV2对象的indices获取。

```python
top_2 = tf.math.top_k(a)
top_2.values, top_2.indices
```

```python
a, tf.math.top_k(tf.transpose(a, [1, 0]), 2) #如果相对其他维度进行部分排序，需要进行维度交换，将需要排序的维度调换到最后
```

### 填充与复制
#### 1）填充 tf.pad()
tf.pad函数主要是用来对tensor的大小进行扩展，包括水平、垂直、深度(通道)等,方法定义如下：

**tf.pad(tensor, paddings, mode="CONSTANT", name=None, constant_values=0)**

输入参数：

- tensor:输入的tensor

- paddings:设置填充的大小

    - 参数paddings必须是形状为（n, 2）的一个list，这里的n是tensor的秩，也就是维度大小。例如当tensor为一个shape为（12，）的tensor时，paddings必须是形如[[x，y]]的一个list，x表示在第一维度前填充值的个数，y表示在第一维度后填充值的个数

- mode:填充方式，默认是CONSTANT,还有REFLECT和SYMMETRIC

- name:名称

- constant_values：CONSTANT填充方式的填充值,默认是0

```python
a = tf.range(1, 13)
a, tf.pad(a, [[3, 0]]) # 第一个维度前面填充3个0， 后面填充0个0（不填充）
```

```python
#当tensor是二维时，paddings必须是shape为（2,2）的list：
a = tf.reshape(a, [3, 4])
a, tf.pad(a, [[1, 1], [3, 0]], constant_values = 3) # 第一维度前后各填充一行，第二维度前填充两行，后不填充，填充值为3
```

```python
#对于3维tensor，paddings是一个shape为（3,2）的list：
a = tf.reshape(a, [2, 2, 3])
a, tf.pad(a, [[1, 0], [1, 1], [1, 0]]) # 第一维度前填充1块数据，后不填充，第二维度前后各填充1行，第三维度前填充1列，后不填充
```

#### 2）pad的使用场景
对于图片数据的高和宽、序列信号的长度，维度长度可能各不相同。为了方便网络的并行计算，需要将不同长度的数据扩张为相同长度，之前我们介绍了通过复制的方式可以增加数据的长度，但是重复复制数据会破坏原有的数据结构，并不适合于此处。通常的做法是，在需要补充长度的信号开始或结束处填充足够数量的特定数值，如 0，使得填充后的长度满足系统要求。那么这种操作就叫做填充(Padding)。

考虑 2 个句子张量，每个单词使用数字编码的方式，如 1 代表 I，2 代表 like 等。第一个句子为：

“I like the weather today.”

我们假设句子编码为：

[1,2,3,4,5,6]

第二个句子为:

“So do I.” 

它的编码为：

[7,8,1,6]

为了能够保存在同一个张量中，我们需要将这两个句子的长度保持一致，也就是说，需要将第二个句子的长度扩充为 6。常见的填充方案是在句子末尾填充若干数量的 0，变成：

[7,8,1,6,0,0]

此时这两个句子堆叠合并 shape 为[2,6]的张量

```python
a = tf.constant([1, 2, 3, 4, 5, 6])
b = tf.constant([7, 8, 1, 6])
b = tf.pad(b, [[0, 2]]) #填充
b
```

填充后句子张量形状一致，再将这 2 句子 Stack 在一起：

```python
tf.stack([a, b], axis = 0)
```

在自然语言处理中，需要加载不同句子长度的数据集，有些句子长度较小，如 10 个单词左右，部份句子长度较长，如超过 100 个单词。为了能够保存在同一张量中，一般会选取能够覆盖大部分句子长度的阈值，如 80 个单词：对于小于 80 个单词的句子，在末尾填充相应数量的 0；对大于 80 个单词的句子，截断超过规定长度的部分单词。以 IMDB 数据集的加载为例：

```python
total_words = 10000 # 设定词汇量大小
max_review_len = 80 # 最大句子长度
embedding_len = 100 # 词向量长度
# 加载 IMDB 数据集
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.imdb.load_data(num_words = total_words)
# 将句子填充或截断到相同长度，设置为末尾填充和末尾截断方式
x_train = tf.keras.preprocessing.sequence.pad_sequences(x_train, maxlen = max_review_len, truncating = 'post', padding = 'post')
x_test = tf.keras.preprocessing.sequence.pad_sequences(x_test, maxlen = max_review_len, truncating = 'post', padding = 'post')
print(x_train.shape, x_test.shape)
```

上述代码中，我们将句子的最大长度 max_review_len 设置为 80 个单词，通过tf.keras.preprocessing.sequence.pad_sequences 可以快速完成句子的填充和截断工作，以其中某个句子为例：

```python
x_train[5]
```

可以看到在句子末尾填充了若干数量的 0，使得句子的长度刚好 80。实际上，也可以选择句子长度不够时，在句子前面填充 0；句子长度过长时，截断句首的单词。经过处理后，所有的句子长度都变为 80，从而训练集可以保存 shape 为[25000,80]的张量，测试集可以保存 shape 为[25000,80]的张量。

考虑对图片的高宽维度进行填充。以 28x28 大小的图片数据为例，如果网络层所接受的数据高宽为 32x32，则必须将 28x28 大小填充到32x32，可以在上、下、左、右方向各填充 2 个单元，如下图所示：



```python
x = tf.random.normal([4,28,28,1])
# 图片上下、左右各填充 2 个单元
tf.pad(x,[[0,0],[2,2],[2,2],[0,0]])#通过填充操作后，图片的大小变为 32x32，满足神经网络的输入要求。
```

#### 3）复制 tf.tile()
tile()方法对指定维度进行复制，定义如下：

tf.tile(input, multiples, name=None):

- input：需要复制的tensor

- multiples: 各维度需要复制的次数，0表示去除数据，1表示不复制，2表示复制一次

    - 参数multiples是一个长度与tensor的秩相等的list，例如当tensor的shape为（12，）时，multiples的shape也必须为只有一个元素的list，

```python
a = tf.range(12)
a, tf.tile(a, [2])
```

```python
#当tensor的shape为（3,4） ndim=2时，multiples是一个包含两个元素的list
a = tf.reshape(a, [3,4])
a,tf.tile(a, [2,3])  # 第一维度复制1次，第二维度复制2次
```

    - value: 要切分的张量
    - num : 要切分的份数，必须与axis指定的轴的长度相等，或者不指定，自动计算
    - axis : 要切分的维度,默认axis=0

```python
tf.unstack(a)
```

```python
tf.unstack(a, axis = 1)
```

```python
tf.unstack(a, axis = 2)
```

#### 1）maxmium()与minmium()
- tf.maximum(tensor,minval):用于限制最小值,也即是说，将一个tensor中小于指定值的元素替换为指定值minval

- tf.minium(tensor,maxval):用于限制一个tensor的最大值，即将tensor中大于指定值的元素替换为指定值maxval

```python
a = tf.range(10)
a,tf.maximum(a, 4), tf.minimum(a, 6)
```

```python
#如果要同时限制一个tensor的最大值和最小值，可以这么做：
tf.minimum(tf.maximum(a, 4), 6)
```

#### 2）clip_by_value()
tf.clip_by_value(tensor,minval,maxval)

clip_by_value()底层也是通过调用minmium()和maxmium()方法来实现同时限制最大值、最小值功能

```python
tf.clip_by_value(a, 4, 6)
```

#### 3）clip_by_norm()
**tf.clip_by_norm(t, clip_norm, axes=None)**

根据tensor的L2范数（模）和给定裁切值按比例对tensor进行限幅。这种方法可以在不改变方向的前提下，按比例对向量进行限幅。

- clip_norm：标量，限定的最大模长

- axes：1-D (tensor)，指明计算norm的维度，默认所有维度参与

```python
a = tf.cast(tf.reshape(tf.range(1, 7), [2, 3]), tf.float32) # 必须是小数
a, tf.norm(a)
```

```python
n5 = tf.clip_by_norm(a, 5) ##限制模长为5, 向量方向不变
n5, tf.norm(n5)
```
