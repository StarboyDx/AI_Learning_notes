### 3.1 文本分类概述
#### 3.1.1 文本分类概念
**定义**

- 文本分类在文本处理中是很重要的一个模块，它的应用也非常广泛，比如：垃圾过滤，新闻分类等等。
- 通常来讲，文本分类任务是指在给定的分类体系中，将文本指定分到某个或某几个类别中。被分类的对象有短文本，例如句子、标题、商品评论等等，长文本，如文章等。分类体系一般人工划分，例如：
    - 1）政治、体育、军事
    - 2）正能量、负能量
    - 3）好评、中性、差评。
- 因此，对应的分类模式可以分为：二分类与多分类问题。

#### 3.1.2 文本分类应用
文本分类的应用十分广泛，可以将其应用在：

- 垃圾邮件的判定：是否为垃圾邮件
- 根据标题为图文视频打标签 ：政治、体育、娱乐等
- 根据用户阅读内容建立画像标签：教育、医疗等
- 电商商品评论分析等等类似的应用：消极、积极
- 自动问答系统中的问句分类

#### 3.1.3 文本分类方法
- 基于机器学习的方法
    - 逻辑回归、朴素贝叶斯、KNN方法、决策树、支持向量机、GBDT/XGBOOST
- 基于深度学习的方法
    - fastText、TextRNN、TextCNN、TextRCNN、TextRNN + Attention\HAN

#### 3.1.4 文本分类建模流程
- 文本分类方法将整个文本分类问题就拆分成了特征工程和分类器两部分。特征工程分为文本预处理、特征提取、文本表示三个部分，最终目的是把文本转换成计算机可理解的格式，并封装足够用于分类的信息，即很强的特征表达能力：



### 3.2 文本分类模型
#### 3.2.1 文本分类模型概述
**文本分类模型分类**

- 机器学习文本分类模型：基于简单机器学习模型构建文本分类模型
- 深度学习文本分类模型：基于RNN、CNN等深度学习模型构建文本分类模型
- 二者在文本特征的使用上也是有区别的

**基于机器学习的文本分类模型**

- 机器学习只能对词向量进行简单的平均或者加权平均
- 模型主要有：逻辑回归、朴素贝叶斯、KNN方法、决策树、支持向量机、GBDT/XGBOOST



**传统机器学习中根据词向量计算文本向量举例**

- 加载预训练词向量
- 分词
- 累加词向量
- 累加词个数
- 求平均值



**基于深度学习的文本分类模型**

- 深度学习是将词向量送入深度学习模型，使用深度学习模型来提取整个文本的表示
- 通过神经网咯可以更好的抓取词语之间的前后语义关系（n-gram）特征
- 深度学习模型主要有：fasttext、TextRNN、TextCNN、TextRCNN、TextRNN + Attention、HAN



#### 3.2.2 文本预处理
**文本预处理的方式**

- 根据前面的简单示例，可以看出由于机器学习和深度学习在文本表示上的不同，导致了二者在文本预处理上也有所不同：（就是多个padding）
- 基于机器学习的文本预处理：tfidf或者word2vec（或glove等）向量加和
- 基于深度学习的文本预处理：word2vec（或glove等）直接输入

**基于机器学习的文本预处理**

- 分词：中文任务分词必不可少，一般使用jieba分词，工业界的翘楚 fool。
- 去停用词：将一些词频较低或者常用的词去掉，减小词典，防止过拟合
- 词性标注：在分词后判断词性（动词、名词、形容词、副词...），词性也是一个特征
- WORD_EMBEDDING：就是word2vec或者glove等。word-embedding作为传统机器学习算法的特征，同时也是深度学习方法必不可少的步骤（深度学习中单字和词的embedding都需要）。
- TFIDF：提取tfidf文本特征

**基于深度学习的文本预处理**

- 分词
- 去停用词
- 创建词典（加入和）
- Padding填充：因为深度学习模型要求输入的文本长度一致
- WORD_EMBEDDING（也可以作为模型部分进行训练）

#### 3.2.4 fastText
**fastText模型优点**

- fastText文本分类原理在第2章已经做了说明，此处不再赘述
- fastText是一个快速的浅层神经网络文本分类算法，与基于深度神经网络的分类算法相比有两大优点：
    - fastText在保持高精度的情况下加快了训练速度和测试速度
    - fastText不需要预训练好的词向量，fastText会自己训练词向量

    

#### 3.2.5 TextRNN
- 利用RNN做文本分类也比较好理解。对于英文，都是基于词的。对于中文，首先要确定是基于字的还是基于词的。如果是基于词，要先对句子进行分词。之后，每个字/词对应RNN的一个时刻，隐层输出作为下一时刻的输入。最后时刻的隐层输出 hT ​抓住整个句子的抽象特征，再接一个softmax进行分类。





#### 3.2.6 TextCNN
**TextCNN的优势**

- 将卷积神经网络CNN应用到文本分类任务，利用多个不同size的kernel来提取句子中的关键信息（类似于多窗口大小的ngram），从而能够更好地捕捉局部相关性。



**TextCNN的详细过程原理**

- Embedding：第一层是图中最左边的7乘5的句子矩阵，每行是词向量，维度=5，这个可以类比为图像中的原始像素点。
- Convolution：然后经过 kernel_sizes=(2,3,4) 的一维卷积层，每个kernel_size 有两个输出 channel。
- MaxPolling：第三层是一个1-max pooling层，这样不同长度句子经过pooling层之后都能变成定长的表示。
- FullConnection and Softmax：最后接一层全连接的 softmax 层，输出每个类别的概率。



#### 3.2.7 TextRCNN
**TextRCNN的原理**

- 一般的 CNN 网络，都是卷积层 + 池化层。这里是将卷积层换成了双向 RNN，所以结果是，两向 RNN + 池化层。就有那么点 RCNN 的味道。



**TextRCNN模型的主要思想**

- 首先获得词向量表示$e(w_i)$
- 其次将词向量通过双向RNN(实践中可以是LSTM或GRU)得到$c_l(w_i)$和$c_r(w_i)$
- 将$c_l(w_i)$，$e(w_i)$以及$c_r(w_i)$拼接得到新的向量，将其输入到全连接网络对其进行整合，激活函数为tanh
- 再将全连接网络的输出进行MaxPooling
- 最后将其输入一个全连接分类器中实现分类



**双向RNN的好处**

- 单向的 RNN（LSTM、GRU），是根据前面的信息推出后面的，但有时候只看前面的词是不够的， 例如：
- 我今天不舒服，我打算__一天。
- 只根据‘不舒服‘，可能推出我打算‘去医院‘，‘睡觉‘，‘请假‘等等
- 但如果加上后面的‘一天‘，能选择的范围就变小了，‘去医院‘这种就不能选了，而‘请假‘‘休息‘之类的被选择概率就会更大。

#### 3.2.8 TextRNN + Attention
**TextRNN + Attention 的由来**

- CNN和RNN用在文本分类任务中尽管效果显著，但都有一个不足的地方就是不够直观，可解释性不好。
- 注意力（Attention）机制是自然语言处理领域一个常用的建模长时间记忆机制，能够很直观的给出每个词对结果的贡献，基本成了Seq2Seq模型的标配。



**TextRNN + Attention 直觉理解**

- 要对一篇文章进行文本分类，人工的做法通常是抓住中心句或者中心词汇，并将其归类。那么，一篇文章中每个句子的重要程度、一个句子中每个词汇的重要程度，都是不一样的。不同的词汇和句子对文章的表达有不同的影响，词汇和句子的重要性是严重依赖于上下文的，即使是相同的词汇和句子，在不同的上下文中重要性也不一样。
- 当我们人对文章进行分类时，通常会将文章分成多个句子，再对每个句子进行分析，这个可以类比于我们平时做英语的阅读理解。

**TextRNN + Attention模型结构**

- word encoder：对词汇进行编码，建立词向量。接着用双向 GRU 从词语的前后两个方向上汇总一个句子的信息来获取词语之间的关系。
- word attention：接着对句子向量使用 Attention 机制，其意义是使用所有词语的信息来表示当前句子，使得句子中包含了整个句子中的重要部分的信息。
- sentence encoder：使用双向 GRU利用上一层表达的句向量再次提取文章的上下文信息。
- sentence attention：对文章向量使用 Attention 机制。
- softmax：常规的输出分类结果。



### 3.3 报警记录分类实战
#### 3.3.1 导入数据

```python
import pandas as pd
data = pd.read_csv("Dataset/baojing/baojing.csv")
data.head()
```

#### 3.3.2 数据预处理

```python
# 导入库
import torch
import jieba
import random
# 处理数据
import numpy as np
#产生同样的结果
SEED = 2019
# Torch
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
#Cuda算法  如果有GPU就设为True
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False
```

```python
from collections import Counter
from torch.utils.data import Dataset

# 分词
data['words'] = data['segment'].apply(lambda x: jieba.lcut(str(x)))

# 构建词频统计与词表
vocab_counter = Counter()
for words in data['words']:
    vocab_counter.update(words)

# 过滤低频词
min_freq = 3
# <pad> 索引为0，<unk> 索引为1
word2idx = {'<pad>': 0, '<unk>': 1}
idx = 2
for word, count in vocab_counter.items():
    if count >= min_freq:
        word2idx[word] = idx
        idx += 1

VOCAB_SIZE = len(word2idx)
print(f"Size of TEXT vocabulary: {VOCAB_SIZE}")
```

```python
unique_labels = data['label'].unique()
label2idx = {label: idx for idx, label in enumerate(unique_labels)}
idx2label = {idx: label for label, idx in label2idx.items()}

NUM_CLASSES = len(label2idx)
print(f"Size of LABEL vocabulary: {NUM_CLASSES}")
print(f"标签映射关系: {label2idx}")
```

```python
# 定义Dataset类
class BaojingDataset(Dataset):
    def __init__(self, dataframe, word2idx, label2idx):
        self.data = dataframe
        self.word2idx = word2idx
        self.label2idx = label2idx

    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        row = self.data.iloc[idx]
        # 文本转索引
        word_indices = [self.word2idx.get(w, 1) for w in row['words']]
        # 标签转整数
        label_str = row['label']
        label_id = self.label2idx[label_str]
        return torch.tensor(word_indices, dtype=torch.long), torch.tensor(label_id, dtype=torch.long)
    
# 划分训练集测试集
train_df = data.sample(frac=0.7, random_state=SEED)
valid_df = data.drop(train_df.index)

train_dataset = BaojingDataset(train_df, word2idx, label2idx)
valid_dataset = BaojingDataset(valid_df, word2idx, label2idx)
```

#### 3.3.3 加载预训练词向量

```python
from gensim.models import KeyedVectors

embedding_dim = 300
# 初始化一个全零或随机权重的矩阵 [VOCAB_SIZE, 300]，后续要有的词替换，没有的词就是随机值，泛化能力更强
embedding_matrix = np.random.normal(scale=0.6, size=(VOCAB_SIZE, embedding_dim))

# 使用 Gensim 加载你下载好的搜狗预训练词向量
# 注意：首次加载较大的 txt 词向量可能需要一点时间
print("loading...")
try:
    sgns_model = KeyedVectors.load_word2vec_format('Dataset/baojing/sgns.sogou.word', binary=False)

    # 将预训练词向量填充到随机矩阵中
    hit_count = 0
    for word, idx in word2idx.items():
        if word in sgns_model:
            embedding_matrix[idx] = sgns_model[word]
            hit_count += 1
    print(f"成功加载预训练向量的词汇数: {hit_count} / {VOCAB_SIZE}")
except FileNotFoundError:
    print("未找到 sgns.sogou.word 文件，将使用随机初始化向量继续。")
```

#### 3.3.4 构建数据集

```python
from torch.utils.data import DataLoader
from torch.nn.utils.rnn import pad_sequence

device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
BATCH_SIZE = 32

# 定义对齐函数
def collate_batch(batch):
    text_list, label_list, length_list = [], [], []
    for _text, _label in batch:
        text_list.append(_text)
        label_list.append(_label)
        length_list.append(len(_text))
    
    labels = torch.tensor(label_list, dtype=torch.long)
    lengths = torch.tensor(length_list, dtype=torch.long)

    # 动态填充，batch_first = True
    padded_texts = pad_sequence(text_list, padding_value=0, batch_first=True)

    # Sort by length (descending) for pack_padded_sequence
    lengths, sorted_idx = lengths.sort(0, descending=True)
    padded_texts = padded_texts[sorted_idx]
    labels = labels[sorted_idx]

    return padded_texts, labels, lengths

# 构建迭代器
train_iterator = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, collate_fn=collate_batch)
valid_iterator = DataLoader(valid_dataset, batch_size=BATCH_SIZE, shuffle=False, collate_fn=collate_batch)
```

#### 3.3.5 搭建模型

```python
import torch.nn as nn
class classifier(nn.Module):   
    #定义所有层
    def __init__(self, vocab_size, embedding_dim, hidden_dim, output_dim, n_layers, 
                 bidirectional, dropout):        
        super().__init__()           
        #embedding 层
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        #lstm 层
        self.lstm = nn.LSTM(embedding_dim, 
                           hidden_dim, 
                           num_layers=n_layers, 
                           bidirectional=bidirectional, 
                           dropout=dropout,
                           batch_first=True)        
        #全连接层
        self.fc = nn.Linear(hidden_dim * 2, output_dim)       
    def forward(self, text, text_lengths):
        #text = [batch size,sent_length]
        embedded = self.embedding(text)#文本嵌入
        #embedded = [batch size, sent_len, emb dim]
        packed_embedded = nn.utils.rnn.pack_padded_sequence(embedded, text_lengths,batch_first=True)   
        packed_output, (hidden, cell) = self.lstm(packed_embedded)
        #hidden = [num layers * num directions,batch size, ,hid dim]
        #cell = [num layers * num directions,batch size,hid dim]  
        #print(hidden.shape)
        #连接最后的正向和反向隐状态  
        hidden = torch.cat((hidden[-2,:,:], hidden[-1,:,:]), dim = 1)                
        #hidden = [batch size, hid dim*2]
        outputs=self.fc(hidden)#全连接输出进行分类  [batch size, output_dim]
        return outputs
```

#### 3.3.6 训练模型

```python
import torch.optim as optim

# 超参数
num_hidden_nodes = 256
num_output_nodes = NUM_CLASSES
num_layers = 2
# bidirection = True
dropout = 0.2

# 实例化模型
model = classifier(VOCAB_SIZE, embedding_dim, num_hidden_nodes, num_output_nodes, num_layers,
                   bidirectional=True, dropout=dropout)

# 把生成的矩阵赋值给 embedding 层
model.embedding.weight.data.copy_(torch.from_numpy(embedding_matrix).float())
print(torch.from_numpy(embedding_matrix).float().shape)
```

```python
model = model.to(device)
optimizer = optim.Adam(model.parameters())
criterion = nn.CrossEntropyLoss().to(device)

#定义度量
def accuracy(preds, y):
    #取最大值的下标的作为预测标签
    rounded_preds = torch.argmax(preds,-1)
    correct = (rounded_preds == y).float()#判断对错 
    acc = correct.sum() / len(correct)#计算准确率
    return acc
```

```python
def train(model, iterator, optimizer, criterion):
    epoch_loss = 0
    epoch_acc = 0
    model.train()

    # 拆包解析 也可以for batch in interator:，
    # 但是这里lengths由于我有gpu，要把这个单独拿到cpu上, 这样更直观
    for text, labels, text_lengths in iterator:
        optimizer.zero_grad()

        text = text.to(device)
        labels = labels.to(device)
        text_lengths = text_lengths.to('cpu')

        # predictions = model(text.T, text_lengths)#模型计算输出值 
        # 取消 .T，因为 batch_first 已经是 True 了
        predictions = model(text, text_lengths)
        loss = criterion(predictions, labels)
        acc = accuracy(predictions, labels)
        # 反向传播更新参数
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()
        epoch_acc += acc.item()
    return epoch_loss / len(iterator), epoch_acc / len(iterator)
```

```python
def evaluate(model, iterator, criterion):  
    #初始化
    epoch_loss = 0
    epoch_acc = 0
    model.eval()#设置为测试模式 停用dropout层
    with torch.no_grad():#测试不需要反向传播  取消autograd
        for text, labels, text_lengths in iterator:
            text = text.to(device)
            labels = labels.to(device)
            text_lengths = text_lengths.to('cpu')

            predictions = model(text, text_lengths)
            #计算损失和准确性
            loss = criterion(predictions, labels)
            acc = accuracy(predictions, labels) 
            #跟踪损失和准确性
            epoch_loss += loss.item()
            epoch_acc += acc.item()
    return epoch_loss / len(iterator), epoch_acc / len(iterator)
```

```python
N_EPOCHS = 20#训练10轮
best_valid_loss = float('inf')#用来保存最优成绩
for epoch in range(N_EPOCHS):   
    #训练模型
    train_loss, train_acc = train(model, train_iterator, optimizer, criterion)  
    #评估模型
    valid_loss, valid_acc = evaluate(model, valid_iterator, criterion) 
    #保存最佳模型
    if valid_loss < best_valid_loss:
        best_valid_loss = valid_loss
        torch.save(model.state_dict(), 'Dataset/baojing/saved_weights.pt')
    #打印训练信息
    print(f'\tTrain Loss: {train_loss:.3f} | Train Acc: {train_acc*100:.2f}%')
    print(f'\t Val. Loss: {valid_loss:.3f} |  Val. Acc: {valid_acc*100:.2f}%')
```
